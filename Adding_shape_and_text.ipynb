#### Adding shape and text
##### In OpenCV,  adding shapes and text means drawing graphical elements (like lines, rectangles, circles) and writing text on an image using built-in drawing functions.

##### This is very important in Biomedical Engineering for:
- Marking tumors in X-ray on MRI.
- Highlighting ECG peaks.
- Displaying patient data on images.

```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
```

```python
img = cv2.imread("highway.jpg", cv2.IMREAD_COLOR)
```

```python
plt.imshow(img)
plt.title("Original Image")
plt.axis("off")
```

##### Line
- cv2.line(image, (x1,y1), (x2,y2), color, thickness)

```python
image_1 = cv2.line(img,(0,0), (150, 150), (255,255,255),5)
image_rgb_1 = cv2.cvtColor(image_1, cv2.COLOR_BGR2RGB)
```

```python
plt.imshow(image_rgb_1)
plt.axis("off")
```

##### Circle
- cv2.circle(image, center_coordinates, radius, color, thickness)

```python
image_2 = cv2.circle(img, (200, 200), 50, (0, 0, 255), 3)
image_rgb_2 = cv2.cvtColor(image_2, cv2.COLOR_BGR2RGB)
```

```python
plt.imshow(image_rgb_2)
plt.axis("off")
```

```python
points = np.array([[100,100], [100,300],[500,50],[500,300]], np.int32)
```

```python
image_3 = cv2.polylines(img, [points], True, (0,255,255), 3)
```

```python
image_rgb_3 = cv2.cvtColor(image_3, cv2.COLOR_BGR2RGB)
```

```python
plt.imshow(image_rgb_3)
plt.axis("off")
```

#### Adding Text
- cv2.putText(image, text, (x,y), font, fontScale, color, thickness)

```python
image_4 = cv2.putText(img, "HELLO", (200, 300), cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 0, 255), 2
```

```python
plt.imshow(image_4)
plt.axis("off")
```
