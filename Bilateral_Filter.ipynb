#### Bilateral Filter
##### Bilateral Filter is an advanced smoothing filter that:
-  Reduces noise
- Preserves edges
- Does not blur edges like Gaussian

##### It is very useful in medical image processing and edge preserving smoothing.

##### Syntax:
- cv2.bilateralFilter(src, d, sigmaColor, sigmaSpace)

where,
- src is a input image.
- d is diameter of the pixel neighborhood.
- sigmaColor is filter sigma in the color space.
- sigmaSpace is filter sigma in the coordinate space.

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
img = cv2.imread("birds.jpg")
```

```python
h, w = img.shape[:2]
print("Original Size", w, "x", h)
```

```python
new_width = 520
new_height = 520
```

```python
resized = cv2.resize(img, (new_width, new_height), interpolation=cv2.INTER_AREA)
```

```python
d = 7
sigmacolor = 500
sigmaspace = 500
```

```python
img_1 = cv2.bilateralFilter(img, d, sigmacolor, sigmaspace)
```

```python
rgb_1 = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
```

```python
rgb_2= cv2.cvtColor(img_1, cv2.COLOR_BGR2RGB)
```

```python
plt.figure(figsize=(8,8))
```

```python
plt.subplot(1,2,1)
plt.imshow(rgb_1)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(rgb_2)
plt.title("Filtered Image")
plt.axis("off")
```
