#### Reading Video
##### A video is a sequence of frames (images) displayed rapidly. In OpenCV, a video is read frame by frame usin the VideoCapture() function.

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
import cv2
import matplotlib.pyplot as plt
from IPython.display import display, clear_output

video = cv2.VideoCapture("video.mp4")

while video.isOpened():
    
    ret, frame = video.read()
    
    if not ret:
        break
    
    # Resize frame
    frame = cv2.resize(frame, (800,720))
    
    # Convert BGR → RGB
    frame = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
    
    # Display frame
    clear_output(wait=True)
    plt.imshow(frame)
    plt.axis("off")
    display(plt.gcf())
    plt.close()

video.release()
```
