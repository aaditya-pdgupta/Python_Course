#### Shifting an image
##### Shifting an image means moving it left, right, up or down without rotating or resizing it. In image processing, this is called Image Translation.

##### cv2.warpAffine()

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
img = cv2.imread("highway.jpg",cv2.IMREAD_COLOR)
h, w = img.shape[:2]
```

```python
tx = 1000
ty = 500
```

```python
T = np.float32([[1, 0, tx], [0, 1, ty]])
```

```python
shifted = cv2.warpAffine(img, T, (w,h))
```

```python
rgb = cv2.cvtColor(shifted, cv2.COLOR_BGR2RGB)
```

```python
plt.figure(figsize=(8,8))
```

```python
plt.subplot(1,2,1)
plt.imshow(img)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(rgb)
plt.title("Shifted Image")
plt.axis("off")
```
