#### Edge Detection
##### Edge detection is a technique in image processing used to identify boundaries of object in an image. Common Edge detection methods are:
##### 1. Canny Edge Detection (Most Popular)
##### Steps in Cannny are :
- Noise reduction (Gaussian filter).
- Gradient calculation.
- Non-maximum suprression
- Double threshold
- Edge Tracking

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
img = cv2.imread("birds.jpg")
```

```python
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
```

```python
resize = cv2.resize(gray, (520,520))
```

```python
edges = cv2.Canny(resize, 100, 200)
```

```python
plt.figure(figsize=(8,8))
```

```python
plt.subplot(1,2,1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(edges, cmap="gray")
plt.title("Edge Detected Image")
plt.axis("off")
```
