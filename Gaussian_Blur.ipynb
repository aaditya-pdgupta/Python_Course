#### Gaussian Blur
##### Gausian Blur is used to smooth an image and reduce noise before operation like:
- Thresholding
-  Edge detection (Canny)
- Contour detection
- Medical image processing

#####  Syntax:
cv2.GaussianBlur(src, ksize,  sigmaX)

where, 
- src is input image
- ksize is kernel size (must be odd,eg (3,3), (5,5))
- sigmaX ---> Vlur strength

```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
import pandas as pd
```

```python
img = cv2.imread("highway.jpg")
gray = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
```

```python
blur = cv2.GaussianBlur(gray, (5,5), 0)
```
