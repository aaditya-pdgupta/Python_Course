#### Image Rotation
##### Rotation is a transformation that turns an amage around a fixed point(usually the center) by a specific angle.

```python
import cv2 
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
img = cv2.imread("highway.jpg",cv2.IMREAD_COLOR)
h, w = img.shape[:2]
```

```python
center = (w//2, h//2)
```

```python
M = cv2.getRotationMatrix2D(center, 45, 1) 
```

```python
rotated = cv2.warpAffine(img, M, (w,h))
```

```python
rgb = cv2.cvtColor(rotated, cv2.COLOR_BGR2RGB)
```

```python
plt.imshow(rgb)
plt.axis("off")
plt.title("Rotated Image")
plt.show()
```
