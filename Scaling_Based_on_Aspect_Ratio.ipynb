```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
```

```python
img = cv2.imread("highway.jpg")
```

```python
h, w = img.shape[:2]
print("Original Image", w, "x", h)
```

```python
new_width = 256
ratio = new_width / w
new_height = int(h * ratio)
```

```python
resized = cv2.resize(img, (new_width, new_height),  interpolation = cv2.INTER_AREA) # Here it is downscaling.
```

```python
resized_rgb =cv2.cvtColor(resized, cv2.COLOR_BGR2RGB)
new_height, new_width = resized.shape[:2]
print("Resized Image",  new_width, "x", new_height)
```

```python
plt.imshow(resized_rgb)
plt.title("Downscaled Image")
plt.axis("off")
```

```python
new_width = 6666
ratio = new_width / w
new_height = int(h * ratio)
```

```python
resized_1 = cv2.resize(img, (new_width, new_height),  interpolation = cv2.INTER_CUBIC) # Here it is the upscaling.
```

```python
resized_rgb_1 =cv2.cvtColor(resized_1, cv2.COLOR_BGR2RGB)
```

```python
new_h, new_w = resized_1.shape[:2]
print("Resized image", new_w, "x", new_h)
```

```python
plt.imshow(resized_rgb_1)
plt.title("Upscaled  Image")
plt.axis("off")
```
