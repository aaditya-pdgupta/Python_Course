#### Thresholding
##### Thresholding is an image processing technique used to convert a grayscale image into a binary image (black and white). It has value from 0 to 255.

##### Syntax 
- ret, thresh = cv2.threshold(src, threshold_value, max_value, type)

```python
import cv2 
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
img = cv2.imread("highway.jpg")
rgb_image = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
```

```python
ret, thresh = cv2.threshold(rgb_image, 127, 255, cv2.THRESH_BINARY)
```
