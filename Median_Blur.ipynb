#### Median Blur
##### Median Blur is a smoothing filter mainly used to remove salt and pepper noise. It replaces each pixel wit the median value of the surrounding pixels.

##### Syntax
-  cv2.medianBlur(src, ksize)

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
img = cv2.imread("highway.jpg")
```

```python
gray = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
```

```python
median = cv2.medianBlur(gray, 5)

plt.imshow(median, cmap='gray')
plt.title("Median Blurred Image")
plt.axis("off")
plt.show()
```
