```python
import cv2
import matplotlib.pyplot as plt
from IPython.display import display, clear_output # Update the frame continuously to create  video effect
 
cap = cv2.VideoCapture(0) # VideoCapture() opens a camera or video file. 0 means default webcam/

while cap.isOpened(): # Checks if the camera is opened successfully.
    
    ret, frame = cap.read() # ret if frame capture succesfully where frame means the actual 
    
    if not ret:
        print("Failed to capture frame")
        break

    # Convert BGR to RGB
    frame_rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)

    plt.imshow(frame_rgb)
    plt.axis('off')

    display(plt.gcf())
    clear_output(wait=True)
    plt.close()

cap.release()
```
